# roof-plants

Each step below is one group of components on a Grasshopper canvas, written as code. A code cell calls
functions and shows the result; the logic lives in `roof_plants/steps.py`, where it is tested and can become
a component later. Change the brief, then rerun from the step that reads it.

Units are millimetres.

```python
%matplotlib inline
import roof_plants as work
```

## Step 1 · The brief
*Canvas: number sliders*

```python
brief = work.Brief(size=3000, cell=900, thickness=120, wear=30, resolution=64)
```

## Step 2 · Envelope and lattice
*Canvas: **Volumetric Box**, **Volumetric TPMS**, **Volumetric Boolean***

```python
envelope = work.steps.envelope(brief)
lattice = work.steps.lattice(brief)
solid = work.steps.porous(envelope, lattice)
work.preview.section(brief, solid, title="porous");
```

## Step 3 · A process
*Canvas: **Volumetric Abrasion***

**Try:** `wear=80` in the brief, and look at where material goes first.

```python
worn = work.steps.weathered(brief, solid)
work.preview.section(brief, worn, title="worn");
```

## Step 4 · Export
*Canvas: **Volumetric Mesh***

Writes the mesh and the field to `out/`, which git ignores. A slicer takes it from there.

```python
vertices, faces = work.steps.export(brief, worn, "out/piece")
print("{:,} triangles".format(len(faces)))
```
